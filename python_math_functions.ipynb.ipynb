```python
#Count how many bits are set to 1 in a number using bitwise operations.
num = int(input("Enter a number: "))
count = 0

while num > 0:
    # Check the last bit
    if num & 1:       # if last bit is 1
        count += 1
    num >>= 1         # right shift by 1 (move to next bit)

print("Number of set bits =", count)
```

```python
#Build a menu-driven calculator that uses arithmetic, comparison, logical, assignment, membership, identity, and bitwise operators wherever applicable.
def calculator():
    num1 = int(input("Enter num1: "))
    num2 = int(input("Enter num2: "))
    print("\n--- Menu ---")
    print("1. Arithmetic Operations")
    print("2. Comparison Operations")
    print("3. Logical Operations")
    print("4. Assignment Operations")
    print("5. Membership Operations")
    print("6. Identity Operations")
    print("7. Bitwise Operations")
    print("8. Exit")  

    while True:
        

        choice = int(input("Enter choice: "))

        if choice == 1:
            print("Addition:", num1 + num2)
            print("Subtraction:", num1 - num2)
            print("Multiplication:", num1 * num2)
            print("Division:", num1 / num2 if num2 != 0 else "Error! Division by zero")
            print("Floor Division:", num1 // num2 if num2 != 0 else "Error! Division by zero")
            print("Modulus:", num1 % num2 if num2 != 0 else "Error! Division by zero")
            print("Exponentiation:", num1 ** num2)

        elif choice == 2:
            print("num1 > num2:", num1 > num2)
            print("num1 < num2:", num1 < num2)
            print("num1 == num2:", num1 == num2)
            print("num1 != num2:", num1 != num2)
            print("num1 >= num2:", num1 >= num2)
            print("num1 <= num2:", num1 <= num2)

        elif choice == 3:
            print("num1 > 0 and num2 > 0:", num1 > 0 and num2 > 0)
            print("num1 > 0 or num2 > 0:", num1 > 0 or num2 > 0)
            print("not(num1 > num2):", not(num1 > num2))

        elif choice == 4:
            a = num1
            print("Initial a =", a)
            a += num2; print("a += num2 →", a)
            a -= num2; print("a -= num2 →", a)
            a *= num2; print("a *= num2 →", a)
            if num2 != 0:
                a //= num2; print("a //= num2 →", a)
                a %= num2; print("a %= num2 →", a)
            a **= 2; print("a **= 2 →", a)

        elif choice == 5:
            s = "python"
            print("'p' in s:", 'p' in s)
            print("'z' not in s:", 'z' not in s)

        elif choice == 6:
            x = [1, 2, 3]
            y = [1, 2, 3]
            z = x
            print("x is y:", x is y)
            print("x is z:", x is z)
            print("x is not y:", x is not y)

        elif choice == 7:
            print("Bitwise AND:", num1 & num2)
            print("Bitwise OR:", num1 | num2)
            print("Bitwise XOR:", num1 ^ num2)
            print("Left shift num1 << 1:", num1 << 1)
            print("Right shift num2 >> 1:", num2 >> 1)

        elif choice == 8:
            print("Exiting calculator...")
            break

        else:
            print("Invalid choice!")

calculator()
```
