Assignment 22: Keyword Arguments

```python
def employee_details(name,id,department,salary):
  print(f"name:{name}")
  print(f"id:{id}")
  print(f"department:{department}")
  print(f"salary:{salary}")
print("before changing the arguements")
employee_details(name="amith",id=101,department="finance",salary=10000)
print("after changing the arguements")
employee_details(id=101,department="finance",name="amith",salary=10000)
# Unlike positional arguments where the order determines what value goes to which variable,
# keyword arguments explicitly specify the variable name during the call (e.g., salary=100000), allowing you to pass them in any order you like without messing up the results.
```

```python
def employee_details(**kwargs):
  print(f"name:{kwargs.get('name')}")
  print(f"id:{kwargs.get('id')}")
  print(f"dept:{kwargs.get('dept')}")
  print(f"salary:{kwargs.get('salary')}")
print("before changing")
employee_details(name="amith" , id=101 ,dept="finance" ,salary=10000)
print("after changing")
employee_details(id=101,dept="finance",name="amith",salary=10000)
```

Assignment 35: Fibonacci Function
Write a function named generate_fibonacci().

Requirements
Accept the number of terms.
Return the complete Fibonacci series as a list.
Handle zero and negative term counts.
Do not use recursion in this assignment.

```python
def generate_fibonacci(n):
  if n<=0:
    return []
  if n==1:
    return [0]
  fib = [0,1]
  for i in range(2,n):
    fib.append(fib[-1]+fib[-2])
  return fib
generate_fibonacci(7)
```

Assignment 75: Filter Dates by Year
Given dates in DD-MM-YYYY format:

Requirements
Ask the user for a year.
Use filter() to return dates belonging to that year.
Handle invalid date strings safely.
Use datetime.strptime() for validation.

```python
from datetime import datetime

dates = [
    "15-08-2024",
    "20-01-2025",
    "31-12-2024",
    "10-05-2023",
    "invalid-date",
    "25-06-2025",
]
year = input("Enter year: ")


def is_valid_and_matches(date_str):

    parts = date_str.split("-")
    if (
        len(parts) == 3
        and all(p.isdigit() for p in parts)
        and len(parts[2]) == 4
    ):
        d, m, y = map(int, parts)
        if 1 <= m <= 12 and 1 <= d <= 31:

            return datetime.strptime(date_str, "%d-%m-%Y").year == int(year)
    return False


result = list(filter(is_valid_and_matches, dates))
print(result)
```

```python
employees = [
    {
        "id": 101,
        "name": "Ravi",
        "department": "IT",
        "salary": 60000,
        "performance_score": 88,
        "experience": 5
    },
    {
        "id": 102,
        "name": "Anitha",
        "department": "HR",
        "salary": 45000,
        "performance_score": 72,
        "experience": 4
    },
    {
        "id": 103,
        "name": "Kiran",
        "department": "IT",
        "salary": 75000,
        "performance_score": 94,
        "experience": 7
    },
    {
        "id": 104,
        "name": "Meena",
        "department": "Finance",
        "salary": 55000,
        "performance_score": 81,
        "experience": 6
    },
    {
        "id": 105,
        "name": "Arjun",
        "department": "IT",
        "salary": 40000,
        "performance_score": 65,
        "experience": 2
    }
]
def display_employees(emp_list):
  if not emp_list:
    print("employees list is not found")
    return
  for emp in emp_list:
    print(f"ID: {emp['id']} | Name: {emp['name']:<8} | Dept: {emp['department']:<8} | Salary: ₹{emp['salary']} | Score: {emp['performance_score']} | Exp: {emp['experience']} yrs")
display_employees(employees)

```

```python
def it_emps(employees):
  it_sector=list(filter(lambda x:x["department"]=="IT",employees))
  return it_sector
result=it_emps(employees)
print(result)
```

```python
def calculate_revised_salaries(emp_list):
    def update_salary(emp):
        updated_emp = emp.copy()
        score = emp["performance_score"]
        if score >= 90:
            hike = 0.15
        elif score >= 80:
            hike = 0.10
        elif score >= 70:
            hike = 0.05
        else:
            hike = 0.00
        updated_emp["revised_salary"] = round(emp["salary"] * (1 + hike), 2)
        return updated_emp
    revised_list = list(map(update_salary, emp_list))
    return revised_list
revised_employees = calculate_revised_salaries(employees)
print("--- Revised Salaries ---")
for emp in revised_employees:
    print(
        f"Name: {emp['name']:<8} | Original Salary: ₹{emp['salary']} | Revised Salary: ₹{emp['revised_salary']}"
    )
```

```python
from functools import reduce
def calculate_total_payroll(emp_list):
  total_payroll=reduce(lambda x,y:x + y["salary"],emp_list,0)
  return total_payroll
total_salary=calculate_total_payroll(employees)
print("total_salary",total_salary)
```

```python
from functools import reduce
def average_score(emp_list):
  if not emp_list:
    return 0.0
  total_score=reduce(lambda x,y:x + y["performance_score"],emp_list,0)
  average=total_score/len(emp_list)
  return round(average,2)
result=average_score(employees)
print(result)
```

```python
def categorize_experience(emp_list):
  def get_band(emp):
    updated_emp=emp.copy()
    exp=emp["experience"]
    if exp >=8:
      band="senior"
    elif exp>=4:
      band="mid-level"
    else:
      band="junior"
    updated_emp["experience_band"]=band
    return updated_emp
  return list(map(get_band,emp_list))
categorized_employees=categorize_experience(employees)
for emp in categorized_employees:
    print(
        f"Name: {emp['name']:<8} | Experience: {emp['experience']} yrs | Band: {emp['experience_band']}"
    )
```

```python
def sort_emps(emp_list):
  sorted_list=sorted(emp_list,key=lambda x:x["performance_score"],reverse=True)
  return sorted_list
sorted_emps=sort_emps(employees)
print(sorted_emps)
```

```python
from functools import reduce
def department_salary(emp_list):
  departments=set(emp["department"] for emp in emp_list)
  dept_salaries={}
  for dept in departments:
    dept_emps=filter(lambda emp:emp["department"]== dept,emp_list)
    total_dept_salary=reduce(lambda x,y:x +y["salary"],dept_emps,0)
    dept_salaries[dept]=total_dept_salary
  return dept_salaries
result=department_salary(employees)
for dept,total in result.items():
  print(f"Department: {dept:<10} | Total Salary: ₹{total:,}")
```
