##LIST OPERATIONS


```python
#Create a list of 5 employee names and add a new employee using append().
emp_names= ["ravi","raju","pinky","ramya","cherry"]
emp_names.append("karunya")
print(emp_names)
```

```python
#Create a list of account numbers and add three new account numbers using extend().
acc_num = ["101","102","103"]
acc_num.extend(["104","105","106"])
print(acc_num)
```

```python
#Insert a new customer at the beginning of a customer list using insert().
emp_names= ["ravi","raju","pinky","ramya","cherry"]
emp_names.insert(0,"ramman")
print(emp_names)
```

```python
#Insert a transaction amount at index 3 in a list.

transaction_amount = ["10000","20000","30000"]
transaction_amount.insert(3,"98700")
print(transaction_amount)
```

```python
#Remove the last item from a list using pop().
emp_names= ["ravi","raju","pinky","ramya","cherry"]

emp_names.pop(-1)
print(emp_names)
```

```python
#Remove the customer "Rahul" from a customer list using remove().
names = ["ramesh","raju","ravikumar","rahul"]
names.remove("rahul")
print(names)
```

```python
#Count how many times the account type "Savings" appears in a list using count().
account_types = ["current","loan","savings","loan","savings","savings"]
result = account_types.count("savings")
print(result)

```

```python
#Find the index of "Current" in a list of account types using index().
account_types = ["current","loan","savings","loan","savings","savings"]
print(account_types.index("current"))
```

```python
#Reverse a list of transaction IDs using reverse()
transaction_ids = ["1234","5748","8253","7587"]
transaction_ids.reverse()
print(transaction_ids)
```

```python
#Sort a list of account balances in ascending order using sort().
amount = ["10000","8947984","30000"]
amount.sort()
print(amount)
```

```python
##Sort a list of customer names in descending order using sort().
customer_names = ["ramya","karunya","cherry","pavithra"]
customer_names.sort(reverse=True)
print(customer_names)
```

```python
#Make a copy of a list using copy() and verify that modifying the copied list does not affect the original list.
abc = ["ravi","123","learning","1239"]
xyz = ["pavithra","fresher","python","23386437"]
abc = xyz.copy()
xyz.append("maddy")
print(abc)
print(xyz)
```

```python
#Clear all elements from a list using clear().
xyz = ["pavithra","fresher","python","23386437"]
xyz.clear()
print(xyz)
```

```python
#Find customers that are present in either customers or new_customers but not both.
customers = {"Rahul", "Anita", "Kiran", "Sneha", "Vikram"}
new_customers = {"Amit", "Priya", "Rahul", "Divya"}
print(customers.symmetric_difference(new_customers))
```
